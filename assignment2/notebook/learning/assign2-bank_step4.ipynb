## 2. Cleaning decisions and target definition

| Item | Decision | Reason |
|---|---|---|
| Target | `y = "yes"` → 1 | Handout positive class |
| `duration` | Dropped | Known only after the call → leakage |
| `pdays = 999` | `pdays_recorded` flag + `pdays = -1` | 999 is a no-contact sentinel, not 999 days; flag = "day count recorded" (some 999 rows have `previous > 0`) |
| `"unknown"` | Own one-hot category | Recorded value with its own signal (checked on train rows, Section 3.1) |
| NaN | No imputation | None present |
| Duplicates | Kept | 12 rows, no ID to confirm |
| `default = "yes"` | Kept | 3 rows; ignored if absent in train |

> **Learning note — why is it OK to clean before splitting?**
> Leakage happens when a step *learns statistics* (a median, a category list, a mean/std) from rows it shouldn't see. Mapping 999 → -1, creating a flag, and dropping a column are fixed rules applied row by row; they learn nothing. Anything learned (encoder categories, scaler mean/std) is fitted after the split, on train only.
>
> **Report prompt (Methods):** In 2–3 sentences, explain why `duration` is excluded and how `pdays = 999` and `"unknown"` are represented.

```python
def prepare_features(raw: pd.DataFrame) -> tuple[pd.DataFrame, pd.Series]:
    """Deterministic, row-wise cleaning only (no statistics learned from the data)."""
    data = raw.copy()
    valid_labels = {"no", POSITIVE_LABEL}
    invalid = ~data[TARGET].isin(valid_labels)
    if invalid.any():
        raise ValueError(f"{TARGET} must be one of {sorted(valid_labels)}; found "
                         f"{data.loc[invalid, TARGET].unique().tolist()} in {int(invalid.sum())} rows")
    y = (data[TARGET] == POSITIVE_LABEL).astype(int).rename("target")

    data["pdays_recorded"] = (data["pdays"] != PDAYS_NO_CONTACT).astype(int)
    data["pdays"] = data["pdays"].where(data["pdays_recorded"] == 1, PDAYS_NO_CONTACT_ENCODED)

    X = data.drop(columns=[TARGET, *LEAKAGE_COLUMNS])
    return X, y


X, y = prepare_features(df)

assert "duration" not in X.columns
assert TARGET not in X.columns
assert (X.loc[X["pdays_recorded"] == 0, "pdays"] == PDAYS_NO_CONTACT_ENCODED).all()

print(f"X: {X.shape}  y: {y.shape}  positive rate: {y.mean():.4f}")
X[["pdays", "pdays_recorded", "previous", "poutcome"]].head()
```

```python
CATEGORICAL_FEATURES = [
    "job", "marital", "education", "default", "housing", "loan",
    "contact", "month", "day_of_week", "poutcome",
]
NUMERIC_FEATURES = [
    "age", "campaign", "pdays", "previous", "pdays_recorded",
    "emp.var.rate", "cons.price.idx", "cons.conf.idx", "euribor3m", "nr.employed",
]

assert set(CATEGORICAL_FEATURES) | set(NUMERIC_FEATURES) == set(X.columns)
assert not set(CATEGORICAL_FEATURES) & set(NUMERIC_FEATURES)
print(f"{len(NUMERIC_FEATURES)} numeric + {len(CATEGORICAL_FEATURES)} categorical = {X.shape[1]} predictors")
```

## 3. Stratified 70/15/15 train / validation / test split

> **Learning note — why three splits, and why two stages?**
> `train_test_split` makes one cut at a time. First 70/30, then split the 30% in half (`test_size = 0.15 / 0.30 = 0.5`).
> - **Train:** fit preprocessing + models
> - **Validation:** early stopping (XGBoost) and choosing configs by validation AP
> - **Test:** touched once at the very end
> `stratify=` keeps the ~11.3% positive rate in every split. Without it, a small split could have noticeably more or fewer positives just by chance.

```python
TRAIN_SIZE, VALID_SIZE, TEST_SIZE = 0.70, 0.15, 0.15

# Stage 1: 70% train vs. 30% holdout. Stage 2: split the holdout 50/50 into validation and test.
X_train, X_hold, y_train, y_hold = train_test_split(
    X, y, train_size=TRAIN_SIZE, stratify=y, random_state=RANDOM_STATE
)
X_valid, X_test, y_valid, y_test = train_test_split(
    X_hold, y_hold, test_size=TEST_SIZE / (VALID_SIZE + TEST_SIZE), stratify=y_hold, random_state=RANDOM_STATE
)

split_summary = pd.DataFrame({
    "rows": [len(y_train), len(y_valid), len(y_test)],
    "share": [len(y_train) / len(y), len(y_valid) / len(y), len(y_test) / len(y)],
    "positives": [y_train.sum(), y_valid.sum(), y_test.sum()],
    "positive_rate": [y_train.mean(), y_valid.mean(), y_test.mean()],
}, index=["train", "validation", "test"]).round(4)
split_summary
```

## 4. Leakage-safe preprocessing

> **Learning note — `fit_transform` vs `transform`**
> - `preprocessor.fit_transform(X_train)` **learns** (category levels, scaler mean/std) from train and applies them.
> - `preprocessor.transform(X_valid / X_test)` **reuses** what was learned; no refitting.
> If the scaler were fitted on all rows, validation and test statistics would leak into training.
>
> **Learning note — why scale for the MLP but not the trees?**
> Trees split on thresholds (`age < 35`), which only depend on order, so scaling changes nothing. Neural nets use gradient descent on weighted sums; features with very different scales (5,000 vs 0/1) make optimization slow or unstable. The handout says to scale for the MLP and lets trees skip it.
>
> **Learning note — "same input information"**
> Both models get the same rows and the same 63 columns. Scaling is a model-specific *representation* change, not new information.

```python
def make_preprocessor(scale_numeric: bool) -> ColumnTransformer:
    numeric_step = StandardScaler() if scale_numeric else "passthrough"
    return ColumnTransformer(
        transformers=[
            ("num", numeric_step, NUMERIC_FEATURES),
            ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), CATEGORICAL_FEATURES),
        ],
        verbose_feature_names_out=False,
    ).set_output(transform="pandas")


def fit_transform_splits(preprocessor: ColumnTransformer):
    """Fit on training rows only; reuse the fitted transformer for validation and test."""
    start = time.perf_counter()
    train_t = preprocessor.fit_transform(X_train)
    valid_t = preprocessor.transform(X_valid)
    test_t = preprocessor.transform(X_test)
    return train_t, valid_t, test_t, time.perf_counter() - start


tree_preprocessor = make_preprocessor(scale_numeric=False)
mlp_preprocessor = make_preprocessor(scale_numeric=True)

X_train_tree, X_valid_tree, X_test_tree, tree_preprocess_sec = fit_transform_splits(tree_preprocessor)
X_train_mlp, X_valid_mlp, X_test_mlp, mlp_preprocess_sec = fit_transform_splits(mlp_preprocessor)

PREPROCESSING_TIME_SEC = {"GBDT (one-hot only)": tree_preprocess_sec, "MLP (one-hot + scaling)": mlp_preprocess_sec}

pd.DataFrame({
    "train": [X_train_tree.shape, X_train_mlp.shape],
    "validation": [X_valid_tree.shape, X_valid_mlp.shape],
    "test": [X_test_tree.shape, X_test_mlp.shape],
    "preprocessing_sec": [round(tree_preprocess_sec, 4), round(mlp_preprocess_sec, 4)],
}, index=["GBDT input", "MLP input"])
```

```python
# 1. Both models see the same encoded columns (same underlying information).
assert list(X_train_tree.columns) == list(X_train_mlp.columns)

# 2. No missing values after preprocessing.
for frame in [X_train_tree, X_valid_tree, X_test_tree, X_train_mlp, X_valid_mlp, X_test_mlp]:
    assert not frame.isna().any().any()

# 3. Tree numeric columns are unchanged; MLP numeric columns are standardized with training statistics.
assert np.allclose(X_train_tree[NUMERIC_FEATURES].to_numpy(), X_train[NUMERIC_FEATURES].to_numpy())

scaling_check = pd.DataFrame({
    "train_mean": X_train_mlp[NUMERIC_FEATURES].mean(),
    "train_std": X_train_mlp[NUMERIC_FEATURES].std(ddof=0),
    "valid_mean": X_valid_mlp[NUMERIC_FEATURES].mean(),
    "test_mean": X_test_mlp[NUMERIC_FEATURES].mean(),
}).round(4)

n_onehot = len(X_train_tree.columns) - len(NUMERIC_FEATURES)
print(f"Encoded features: {X_train_tree.shape[1]} ({len(NUMERIC_FEATURES)} numeric + {n_onehot} one-hot)")
print("Categories with no training rows (ignored if seen later):",
      [f"{col}={val}" for col in CATEGORICAL_FEATURES for val in X[col].unique()
       if val not in set(X_train[col])] or "none")
scaling_check
```

## 5. Gradient-boosted decision trees (XGBoost)

All runs use `xgboost.XGBClassifier(tree_method="hist")` on the GBDT input from Section 4 (one-hot, unscaled numerics), the same training rows, and the same validation rows.

**Fixed settings**

- `n_estimators = 5000` as an upper limit, with `early_stopping_rounds = 50` and `eval_metric = "logloss"` set in the constructor.
- `eval_set = [(train), (validation)]`: binary log loss is tracked on both, and validation, listed last, controls early stopping.
- `random_state = 42`; all other hyperparameters at XGBoost defaults.

**Two different metrics.** Validation *log loss* decides when to stop adding trees within a run. Validation *Average Precision* (from `predict_proba`, via `average_precision_score`) compares configurations and selects the final model. The test set is not used here.

**Controlled experiments (8 configurations)**

| Phase | Varied | Held fixed |
|---|---|---|
| A | `learning_rate` ∈ {0.01, 0.1, 0.3} | `max_depth = 4`, `subsample = 1.0` |
| B | `max_depth` ∈ {2, 4, 6, 8} | best learning rate from A, `subsample = 1.0` |
| C | `subsample` ∈ {1.0, 0.8, 0.6} (regularization) | best learning rate and depth |

Phases share their reference configuration (for example, depth 4 in phase B is the phase A run), so each comparison changes one setting at a time.

> **Learning note — what gradient boosting does**
> Boosting builds trees one after another. Each new tree is fitted to the *mistakes* (gradients of the log loss) of the trees so far, and its prediction is added in, scaled by `learning_rate`. After many rounds the sum of small trees is a strong model.
>
> **Learning note — `hist`**
> `tree_method="hist"` buckets each numeric feature into bins before searching for splits. It's much faster than checking every unique value and is what the handout requires.
>
> **Learning note — early stopping, step by step**
> 1. After each new tree, XGBoost computes log loss on every set in `eval_set`.
> 2. The **last** set (validation) is watched.
> 3. If validation log loss hasn't improved for 50 rounds, training stops; `best_iteration` records the round with the lowest validation loss.
> 4. `predict_proba` then uses only the trees up to `best_iteration`.
> So `n_estimators=5000` is a ceiling, not the number of trees actually used. The handout wants `early_stopping_rounds` and `eval_metric` in the **constructor** (newer XGBoost versions removed them from `fit()`).
>
> **Learning note — log loss vs AP (common review question)**
> - **Log loss** checks how good the predicted probabilities are, round by round; that's what early stopping needs.
> - **AP** is the assignment's primary metric and is used to **choose between configurations**.
> They answer different questions, so using both is not a contradiction.

```python
GBDT_N_ESTIMATORS = 5000          # upper limit; early stopping decides the actual number of trees
GBDT_EARLY_STOPPING_ROUNDS = 50   # stop after 50 rounds without validation log-loss improvement
GBDT_BASE_PARAMS = {"learning_rate": 0.1, "max_depth": 4, "subsample": 1.0}

gbdt_models: dict[str, xgb.XGBClassifier] = {}
gbdt_records: list[dict] = []


def fit_gbdt(config_id: str, phase: str, **overrides) -> dict:
    """Train one XGBoost configuration with validation-based early stopping and record the result."""
    params = {**GBDT_BASE_PARAMS, **overrides}
    model = xgb.XGBClassifier(
        tree_method="hist",
        n_estimators=GBDT_N_ESTIMATORS,
        early_stopping_rounds=GBDT_EARLY_STOPPING_ROUNDS,
        eval_metric="logloss",
        random_state=RANDOM_STATE,
        n_jobs=-1,
        **params,
    )
    start = time.perf_counter()
    # Validation is listed last, so it controls early stopping.
    model.fit(X_train_tree, y_train, eval_set=[(X_train_tree, y_train), (X_valid_tree, y_valid)], verbose=False)
    fit_sec = time.perf_counter() - start

    history = model.evals_result()
    best = model.best_iteration
    trees_built = len(history["validation_1"]["logloss"])
    train_loss = history["validation_0"]["logloss"][best]
    valid_loss = history["validation_1"]["logloss"][best]
    # With early stopping, predict_proba uses the trees up to best_iteration.
    valid_ap = average_precision_score(y_valid, model.predict_proba(X_valid_tree)[:, 1])

    record = {
        "config": config_id,
        "phase": phase,
        **params,
        "best_iteration": best,
        "trees_built": trees_built,
        # Cap-limited only if the run ended at the cap before early stopping could trigger.
        "hit_cap": trees_built == GBDT_N_ESTIMATORS and trees_built - 1 - best < GBDT_EARLY_STOPPING_ROUNDS,
        "train_logloss": train_loss,
        "valid_logloss": valid_loss,
        "logloss_gap": valid_loss - train_loss,
        "valid_AP": valid_ap,
        "fit_sec": fit_sec,
    }
    # Re-running a phase replaces its earlier record and model under the same ID.
    gbdt_records[:] = [r for r in gbdt_records if r["config"] != config_id]
    gbdt_models[config_id] = model
    gbdt_records.append(record)
    return record


def gbdt_table(records: list[dict]) -> pd.DataFrame:
    return pd.DataFrame(records).set_index("config").round(4)
```

> **Reading the code**
> - `GBDT_BASE_PARAMS` is the reference configuration; each phase overrides one setting.
> - `fit_gbdt` trains, times, and records one run; every run is kept in `gbdt_records` so the table shows **all** configurations tried, as the handout requires.
> - `logloss_gap = valid − train` at the best iteration: a bigger gap means more overfitting.
> - `hit_cap = True` would mean training reached `n_estimators` trees **and** the best iteration was within the last 50 rounds, so early stopping could not have triggered, i.e. the run wasn't converged and the cap should be raised. (An earlier 3,000-tree cap was hit by depth 2 at lr 0.01, which is why the cap is now 5,000.)

### 5.1 Phase A — learning rate

> **Learning note — learning rate (shrinkage)**
> Each tree's contribution is multiplied by `learning_rate`. Small values = small, careful steps → many trees, smoother curves, slower training. Large values = big steps → few trees, faster, but it can overshoot and overfit sooner.
> **Prediction before running:** 0.01 should need the most trees; 0.3 the fewest.

```python
for i, lr in enumerate([0.01, 0.1, 0.3], start=1):
    fit_gbdt(f"G{i}", "A: learning rate", learning_rate=lr)

phase_a = [r for r in gbdt_records if r["phase"] == "A: learning rate"]
best_lr = max(phase_a, key=lambda r: r["valid_AP"])["learning_rate"]
print(f"Learning rate carried forward (best validation AP in phase A): {best_lr}")
gbdt_table(phase_a)
```

### 5.2 Phase B — tree depth

> **Learning note — tree depth = model complexity**
> A depth-*d* tree can model interactions among up to *d* features. Shallow trees (2) are simple → may **underfit** (high bias). Deep trees (8) can memorize training quirks → **overfit** (high variance). Watch `train_logloss` vs `valid_logloss`: the gap should grow with depth.

```python
for i, depth in enumerate([2, 6, 8], start=4):
    fit_gbdt(f"G{i}", "B: max depth", learning_rate=best_lr, max_depth=depth)

# Depth 4 at the chosen learning rate was already trained in phase A and is reused as the reference.
depth_comparison = [r for r in gbdt_records if r["learning_rate"] == best_lr and r["subsample"] == 1.0]
best_depth = max(depth_comparison, key=lambda r: r["valid_AP"])["max_depth"]
print(f"Max depth carried forward (best validation AP among depths): {best_depth}")
gbdt_table(sorted(depth_comparison, key=lambda r: r["max_depth"]))
```

### 5.3 Phase C — row subsampling

`subsample` trains each tree on a random fraction of the training rows. It is the regularization control varied here: adding randomness makes individual trees less able to fit noise in the training set.

> **Learning note — why `subsample` counts as regularization**
> Regularization = anything that discourages the model from fitting noise. With `subsample=0.6`, each tree sees a different random 60% of the rows, so no single tree can lock onto quirks of particular rows. It's similar in spirit to bagging. (`reg_alpha` / `reg_lambda` would instead penalize large leaf weights; the handout lets us pick one control.)

```python
for i, subsample in enumerate([0.8, 0.6], start=7):
    fit_gbdt(f"G{i}", "C: subsample", learning_rate=best_lr, max_depth=best_depth, subsample=subsample)

# subsample = 1.0 at the chosen learning rate and depth is the reference from earlier phases.
subsample_comparison = [r for r in gbdt_records if r["learning_rate"] == best_lr and r["max_depth"] == best_depth]
gbdt_table(sorted(subsample_comparison, key=lambda r: -r["subsample"]))
```

### 5.6 Feature importance

Importance measure: **total gain**, the sum of improvements in the training objective from splits on a feature across the trees used for prediction (boosting rounds `0` through `best_iteration`, inclusive), shown as a share of that total.
```python
# Importance over trees used for prediction only: Tree index is 0-based and
# best_iteration is inclusive, matching predict_proba under early stopping.
# Use total_gain (sum of split gains), not average gain per split.
booster = gbdt_selected.get_booster()
best_it = gbdt_selected.best_iteration
trees = booster.trees_to_dataframe()
split_gains = trees.loc[(trees["Tree"] <= best_it) & (trees["Feature"] != "Leaf")]
gain = split_gains.groupby("Feature")["Gain"].sum()
gain_share = (gain / gain.sum()).sort_values(ascending=False)
TOP_N = 15

fig, ax = plt.subplots(figsize=(8, 6))
gain_share.head(TOP_N).iloc[::-1].plot.barh(ax=ax, color="tab:blue")
ax.set_xlabel("Share of total gain")
ax.set_title(f"Selected GBDT ({GBDT_SELECTED_ID}): top {TOP_N} features by total gain")
ax.grid(axis="x", alpha=0.3)
fig.tight_layout()
fig.savefig("fig_gbdt_feature_importance.png", dpi=200, bbox_inches="tight")
plt.show()

print(f"Trees included: 0..{best_it} (best_iteration inclusive); "
      f"booster also has {booster.num_boosted_rounds() - (best_it + 1)} early-stopping extra rounds not used here")
print(f"Features used in at least one split: {len(gain)} of {X_train_tree.shape[1]}")
print(f"Top feature share: {gain_share.index[0]} = {gain_share.iloc[0]:.4f}")
gain_share.head(TOP_N).round(4).to_frame("gain_share")

```

> - **weight**: how many times the feature was used in a split; it favors features with many possible split points.
> - **cover**: how many rows were affected by those splits.

## 6. Multi-layer perceptron (scikit-learn)

All runs use `sklearn.neural_network.MLPClassifier(solver="adam")` on the MLP input from Section 4 (one-hot categoricals plus numeric features standardized with training statistics), the same training rows, and the same validation rows as the GBDT experiments.

**Fixed settings**

- `max_iter = 200` epochs (scikit-learn's default) as the training budget, identical for every configuration.
- `early_stopping = False`, as recommended in the handout: all training rows are used for weight updates, and the external validation set is used only to compare configurations. Training can still end before `max_iter`: scikit-learn stops when the training loss improves by less than `tol = 1e-4` for 10 consecutive epochs, which counts as converged. A run that uses the full budget without meeting that criterion raises a `ConvergenceWarning`; these warnings are captured, printed, and recorded per run in the `converged` column.
- `activation = "relu"`, `batch_size = "auto"` (200 rows), `random_state = 42`; all other hyperparameters at scikit-learn defaults.

**What is measured.** `loss_curve_` is the training loss per epoch; it includes the L2 penalty, so its level depends on `alpha`. After training, `train_logloss` and `valid_logloss` are computed from `predict_proba` without the penalty, which makes them comparable across configurations and with the GBDT table. Validation Average Precision selects the final configuration. The test set is not used here.

**Controlled experiments (8 configurations, the same number as for GBDT)**

| Phase | Varied | Held fixed |
|---|---|---|
| A | `hidden_layer_sizes` ∈ {(64,), (128, 64), (128, 64, 32)} | `learning_rate_init = 0.001`, `alpha = 0.0001` (scikit-learn defaults) |
| B | `learning_rate_init` ∈ {0.0001, 0.001, 0.01} | best architecture from A, `alpha = 0.0001` |
| C | `alpha` ∈ {0.0001, 0.01, 0.1, 1.0} (L2 regularization) | best architecture and learning rate |

As for GBDT, phases share their reference configuration, so each comparison changes one setting at a time.

> **Learning note — what an MLP does**
> Each hidden layer computes `relu(W·x + b)`: a weighted sum of the previous layer followed by a non-linearity. Stacking layers lets the network learn interactions between features. The output layer squashes one number through a sigmoid to get P(subscribe).
>
> **Learning note — why the MLP needs scaled inputs but trees don't**
> Gradient descent updates every weight with the same step size. If `nr.employed` is ~5,000 and `pdays_recorded` is 0/1, the gradients for those weights differ by orders of magnitude and training is slow or unstable. Trees only ask "is x < threshold?", so scaling doesn't change their splits. The scaler was fitted on training rows only (Section 4) — no leakage.
>
> **Learning note — Adam**
> Adam is gradient descent with per-weight adaptive step sizes and momentum. `learning_rate_init` is its base step size. The handout requires `solver="adam"`.
>
> **Learning note — what "converged" means in scikit-learn**
> With `early_stopping=False`, scikit-learn watches the **training** loss. If it improves by less than `tol` (1e-4) for `n_iter_no_change` (10) epochs in a row, training stops and the model counts as converged. If it reaches `max_iter` first, it raises a `ConvergenceWarning`. "Converged" therefore means "training loss flattened", **not** "best validation performance" — a converged network can still be badly overfit.
>
> **Learning note — why `early_stopping=False`**
> With `early_stopping=True`, scikit-learn would hold out 10% of the *training* rows as its own internal validation set. That changes the effective training data and adds a second validation set alongside ours. The handout recommends `False` for a simple comparison; the cost is that nothing stops the network from overfitting except the epoch budget and `alpha`.

```python
MLP_MAX_ITER = 200   # epoch budget, identical for every configuration
MLP_BASE_PARAMS = {"hidden_layer_sizes": (64,), "learning_rate_init": 0.001, "alpha": 0.0001}

mlp_models: dict[str, MLPClassifier] = {}
mlp_records: list[dict] = []


def fit_mlp(config_id: str, phase: str, **overrides) -> dict:
    """Train one MLP configuration for the fixed epoch budget and record the result."""
    params = {**MLP_BASE_PARAMS, **overrides}
    model = MLPClassifier(
        solver="adam",
        activation="relu",
        max_iter=MLP_MAX_ITER,
        early_stopping=False,
        random_state=RANDOM_STATE,
        **params,
    )
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always", ConvergenceWarning)
        start = time.perf_counter()
        model.fit(X_train_mlp, y_train)
        fit_sec = time.perf_counter() - start

    # Report convergence warnings explicitly and pass any other warning through unchanged.
    convergence_warnings = [w for w in caught if issubclass(w.category, ConvergenceWarning)]
    for w in caught:
        if issubclass(w.category, ConvergenceWarning):
            print(f"{config_id}: ConvergenceWarning — {w.message}")
        else:
            warnings.warn_explicit(w.message, w.category, w.filename, w.lineno)

    train_proba = model.predict_proba(X_train_mlp)[:, 1]
    valid_proba = model.predict_proba(X_valid_mlp)[:, 1]
    train_loss = log_loss(y_train, train_proba)
    valid_loss = log_loss(y_valid, valid_proba)

    record = {
        "config": config_id,
        "phase": phase,
        **params,
        "epochs": model.n_iter_,
        "converged": not convergence_warnings,
        "final_train_loss": model.loss_curve_[-1],
        "train_logloss": train_loss,
        "valid_logloss": valid_loss,
        "logloss_gap": valid_loss - train_loss,
        "valid_AP": average_precision_score(y_valid, valid_proba),
        "fit_sec": fit_sec,
    }
    # Re-running a phase replaces its earlier record and model under the same ID.
    mlp_records[:] = [r for r in mlp_records if r["config"] != config_id]
    mlp_models[config_id] = model
    mlp_records.append(record)
    return record


def mlp_table(records: list[dict]) -> pd.DataFrame:
    return pd.DataFrame(records).set_index("config").round(4)
```

> **Reading the code**
> - `MLP_BASE_PARAMS` is the reference configuration (scikit-learn's defaults for learning rate and `alpha`); each phase overrides one setting.
> - `fit_mlp` trains, times, and records one run; all runs stay in `mlp_records`, so the table shows every configuration tried.
> - `warnings.catch_warnings(record=True)` **captures** warnings so we can print them next to the run ID and store `converged`. It does not hide them; other warning types are re-raised unchanged.
> - `epochs` = `n_iter_`: how many passes over the training rows actually ran (≤ 200).
> - `final_train_loss` is the last value of `loss_curve_` and **includes** the L2 penalty; `train_logloss` / `valid_logloss` are plain log loss, comparable to the GBDT table.
> - `logloss_gap = valid − train`: a large gap means overfitting, same meaning as for GBDT.

### 6.1 Phase A — architecture

Three architectures of increasing width and depth: one hidden layer of 64 units, two layers (128, 64), and three layers (128, 64, 32).

> **Learning note — width and depth = capacity**
> Number of trainable weights with 63 inputs:
> - `(64,)`: 63·64 + 64 + 64 + 1 = **4,161**
> - `(128, 64)`: 8,192 + 8,256 + 65 = **16,513**
> - `(128, 64, 32)`: 8,192 + 8,256 + 2,080 + 33 = **18,561**
>
> More weights = more capacity to fit the ~28,800 training rows, including their noise. With a fixed 200-epoch budget, low `alpha`, and no early stopping, bigger networks can memorize the training set: watch for a small `train_logloss` paired with a large `valid_logloss`.

```python
for i, layers in enumerate([(64,), (128, 64), (128, 64, 32)], start=1):
    fit_mlp(f"M{i}", "A: architecture", hidden_layer_sizes=layers)

phase_a_mlp = [r for r in mlp_records if r["phase"] == "A: architecture"]
best_layers = max(phase_a_mlp, key=lambda r: r["valid_AP"])["hidden_layer_sizes"]
print(f"Architecture carried forward (best validation AP in phase A): {best_layers}")
mlp_table(phase_a_mlp)
```

### 6.2 Phase B — initial learning rate

`learning_rate_init` is Adam's base step size. The 0.001 run at the chosen architecture is the reference from phase A.

> **Learning note — learning rate for neural networks**
> Too large: the loss jumps around or the network fits the training data fast and starts overfitting early. Too small: training is slow and may hit `max_iter` before converging (→ `ConvergenceWarning`). Unlike GBDT, there is no early stopping here, so the learning rate also decides **how far** the network gets toward overfitting within 200 epochs. Compare with GBDT, where early stopping made learning rate mostly a speed knob.

```python
for i, lr in enumerate([0.0001, 0.01], start=4):
    fit_mlp(f"M{i}", "B: learning rate", hidden_layer_sizes=best_layers, learning_rate_init=lr)

lr_comparison = [r for r in mlp_records
                 if r["hidden_layer_sizes"] == best_layers and r["alpha"] == MLP_BASE_PARAMS["alpha"]]
best_lr_init = max(lr_comparison, key=lambda r: r["valid_AP"])["learning_rate_init"]
print(f"Learning rate carried forward (best validation AP in phase B): {best_lr_init}")
mlp_table(sorted(lr_comparison, key=lambda r: r["learning_rate_init"]))
```

### 6.3 Phase C — L2 regularization (`alpha`)

`alpha` scales an L2 penalty on the network weights that is added to the training loss. Larger values push weights toward zero and limit how closely the network can fit the training rows.

> **Learning note — `alpha` (L2 regularization)**
> Training minimizes `log loss + (alpha / 2n) · Σ w²`. Larger `alpha` = smaller weights = smoother decision function = less overfitting, but too large underfits. The handout lets us vary **one** of `activation` or `alpha`; `alpha` is the direct counterpart of GBDT's regularization phase (`subsample`).
>
> Note: because `loss_curve_` includes the penalty, curves with different `alpha` sit at different levels even if their predictions are equally good. Compare configurations with `valid_logloss` / `valid_AP`, not with the final `loss_curve_` value.

```python
for i, alpha in enumerate([0.01, 0.1, 1.0], start=6):
    fit_mlp(f"M{i}", "C: L2 alpha", hidden_layer_sizes=best_layers, learning_rate_init=best_lr_init, alpha=alpha)

# alpha = 0.0001 at the chosen architecture and learning rate is the reference from earlier phases.
alpha_comparison = [r for r in mlp_records
                    if r["hidden_layer_sizes"] == best_layers and r["learning_rate_init"] == best_lr_init]
mlp_table(sorted(alpha_comparison, key=lambda r: r["alpha"]))
```

## 7. Final model comparison and one-time test evaluation

Both final configurations were fixed by validation Average Precision before this section: the GBDT in Section 5.4 and the MLP in Section 6.4. Nothing below changes those choices. The test set is used here for the first and only time.

**Metrics.** Average Precision is computed from positive-class probabilities with `average_precision_score`. Accuracy, precision, recall, and F1 for the positive class use a fixed threshold of 0.5 (a client is predicted to subscribe when the predicted probability is at least 0.5). No threshold is tuned.

**Timing.** Both models are timed on the same Colab CPU runtime. Each selected configuration is refitted `TIMING_REPEATS = 3` times and the median is reported, because single timings on a shared runtime are noisy (Sections 5–6 show this). Preprocessing time is measured separately, also as the median of three runs, and covers fitting the preprocessor on the training rows and transforming training, validation, and test rows.

> **Learning note — why the test set is used only now**
> Every choice (cleaning, features, hyperparameters, which configuration wins) was made with training and validation data. If we looked at test results and then changed anything, the test score would no longer be an honest estimate of performance on new clients. That's why the handout says "evaluate once".
>
> **Learning note — AP vs threshold metrics**
> - **AP** summarises the whole precision–recall curve from probabilities; it doesn't depend on any threshold. It's the primary metric.
> - **Accuracy / precision / recall / F1** need hard yes/no predictions, so they depend on the threshold. The handout fixes it at 0.5.
> - With an 11% positive rate, 0.5 is a high bar: the model must be quite sure before saying "yes", so expect high precision but low recall for both models.

### 7.2 Training and preprocessing time

The GBDT refit passes only the validation set to `eval_set`, which is all early stopping needs; the training-set log loss tracked in Section 5 was for the learning-curve plots. XGBoost uses all CPU cores (`n_jobs=-1`); the MLP uses NumPy's multithreaded linear algebra. Each timed refit is checked to reproduce the selected model exactly, so the timings describe the models evaluated on the test set.

```python
TIMING_REPEATS = 3


def timed_refits(estimator, X_fit, fit_kwargs: dict) -> tuple:
    """Refit a fresh copy of `estimator` TIMING_REPEATS times; return the last model and all fit times."""
    times = []
    for _ in range(TIMING_REPEATS):
        model = clone(estimator)
        start = time.perf_counter()
        model.fit(X_fit, y_train, **fit_kwargs)
        times.append(time.perf_counter() - start)
    return model, times


gbdt_refit, gbdt_fit_times = timed_refits(
    gbdt_selected, X_train_tree, {"eval_set": [(X_valid_tree, y_valid)], "verbose": False})
mlp_refit, mlp_fit_times = timed_refits(mlp_selected, X_train_mlp, {})

assert gbdt_refit.best_iteration == gbdt_selected.best_iteration
assert np.allclose(gbdt_refit.predict_proba(X_valid_tree), gbdt_selected.predict_proba(X_valid_tree))
assert np.allclose(mlp_refit.predict_proba(X_valid_mlp), mlp_selected.predict_proba(X_valid_mlp))

preprocess_times = {"GBDT (XGBoost)": [], "MLP": []}
for _ in range(TIMING_REPEATS):
    *_, sec = fit_transform_splits(make_preprocessor(scale_numeric=False))
    preprocess_times["GBDT (XGBoost)"].append(sec)
    *_, sec = fit_transform_splits(make_preprocessor(scale_numeric=True))
    preprocess_times["MLP"].append(sec)

timing = pd.DataFrame({
    "preprocessing_sec (median)": {name: np.median(t) for name, t in preprocess_times.items()},
    "fit_sec (median)": {"GBDT (XGBoost)": np.median(gbdt_fit_times), "MLP": np.median(mlp_fit_times)},
    "fit_sec (min)": {"GBDT (XGBoost)": min(gbdt_fit_times), "MLP": min(mlp_fit_times)},
    "fit_sec (max)": {"GBDT (XGBoost)": max(gbdt_fit_times), "MLP": max(mlp_fit_times)},
}).round(3)
print(f"Fit times over {TIMING_REPEATS} refits on {os.cpu_count()} CPU cores "
      f"({platform.platform()}); refits reproduce the selected models.")
timing
```

> **Learning note — what makes this timing fair**
> - Same machine/runtime, same session, same training rows, CPU only for both.
> - Median of 3 refits, because one timing on Colab's shared CPUs can be off by 2× (compare `fit_sec` of similar GBDT configs in Section 5).
> - **Fit time** = `model.fit(...)` only. **Preprocessing time** = fitting the encoder/scaler on train + transforming all three splits. They are separate numbers, as the handout requires.
> - The asserts prove the timed refit produces the same model as the selected one, so we aren't timing something different from what we evaluate.
> - Caveat to mention in the report: the times measure *one final fit*, not the whole search (8 configs each); the GBDT also includes per-round validation scoring for early stopping.
